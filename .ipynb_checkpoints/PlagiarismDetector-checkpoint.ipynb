```python
import os
import pandas as pd
from transformers import AutoTokenizer

# -----------------------------
# Paths
# -----------------------------
base_path = r"C:\Users\majos\Downloads\SOCO"

# -----------------------------
# Leer CSV
# -----------------------------
train_df = pd.read_csv(
    os.path.join(base_path, "train.csv")
)

# -----------------------------
# Guardar códigos y labels
# -----------------------------
codigos = []
labels = []

for _, row in train_df.iterrows():

    nombre_archivo = str(row["pid"])

    ruta_archivo = os.path.join(
        base_path,
        "train",
        nombre_archivo
    )

    try:
        with open(ruta_archivo, "r", encoding="utf-8") as f:

            codigo = f.read()

        codigos.append(codigo)

        # uid = label
        labels.append(row["uid"])

    except Exception as e:
        print(f"Error leyendo {ruta_archivo}: {e}")

# -----------------------------
# Tokenizer CodeBERT
# -----------------------------
tokenizer = AutoTokenizer.from_pretrained(
    "microsoft/codebert-base"
)

tokens = tokenizer(
    codigos,
    padding=True,
    truncation=True,
    max_length=512,
    return_tensors="pt"
)

print(tokens.keys())
print("input ids" + tokens["input_ids"].shape)
print(labels[:5])
```
